## Day 1 — Assets, Threats & Risk

### Learning objective
Separate **asset**, **threat**, **vulnerability**, **control**, **likelihood** and **impact**.

**Analogy:** A bank vault is an asset-protection system. The wall is only one control; identity checks, dual authorization, cameras, alarms, logs and incident response matter too.

### Practice
Create at least 8 assets for a fictional enterprise GenAI platform. Score likelihood and impact from 1–5. Risk score = likelihood × impact.

```python
assets = [
    {"asset":"LLM API endpoint","likelihood":3,"impact":4},
    {"asset":"Identity token","likelihood":4,"impact":5},
    {"asset":"Vector database","likelihood":3,"impact":5},
    {"asset":"CI/CD pipeline","likelihood":3,"impact":5},
]
for a in assets:
    a["risk_score"] = a["likelihood"] * a["impact"]
sorted(assets, key=lambda x:x["risk_score"], reverse=True)
```

## Day 3 — Identity as the New Perimeter

### Learning objective
Recognize that compromised identity can bypass otherwise strong network controls.

### Practice
Review fictional permissions. Flag wildcard or administrator-style privileges and propose narrower alternatives.

```python
permissions = {
 "genai-app":["model:invoke","storage:read"],
 "build-runner":["*"],
 "analyst":["logs:read"],
 "ops-admin":["admin:*"]
}
for principal, perms in permissions.items():
    risky = any("*" in p for p in perms)
    print(f"{principal:12} | {'REVIEW' if risky else 'bounded'} | {perms}")
```
